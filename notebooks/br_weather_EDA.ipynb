```python
import pandas as pd
df = pd.read_csv("../data_brazilian/southeast.csv")
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
df = df.replace(-9999, pd.NA)
df = df.replace(-9999.0, pd.NA)
```

```python
df['timestamp'] = pd.to_datetime(df['Data'] + ' ' + df['Hora'], errors='coerce')
df_final = df[(df['timestamp'] >= '2020-01-01') & (df['timestamp'] <= '2021-12-31')]
print(df_final.shape)
```

```python
df_final.head()
```

```python
stations = df_final['station_code'].unique()
print(len(stations))
print(df_final.dtypes)
```

```python
weather_cols = [c for c in df_final.columns if c not in ['index','height', 'Data', 'Hora', 'timestamp', 
                                                             'region', 'state', 'station', 
                                                             'station_code', 'latitude', 'longitude', 'height']]

for col in weather_cols:
    df_final[col] = df_final[col].astype(str).str.replace(',', '.', regex=False)
    df_final[col] = pd.to_numeric(df_final[col], errors='coerce')
```
